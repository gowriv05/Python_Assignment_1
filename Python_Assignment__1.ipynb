Objective:
To concatenate two strings and display the result.

The following methods are used:

* `upper()`
* `lower()`

```python
# 3. String Methods
strM="Python beginner tutorial"

# a. Convert the sentence to uppercase
print(strM.upper())

# b. Convert the sentence to lowercase
print(strM.lower())

# c. Use Capitalize and return the sentence to the original input form
print(strM.capitalize())

# d. Count the total number of occurrences of character ‘t’ in the string
print(strM.count("t"))

# e. Replace all occurrences of “Python” with “Data Analytics” in the input stringstrM = “Python beginner tutorial.”
print(strM.replace("Python","Data Analytics"))
```

4.Tuple Operations

Objective

To create tuples and perform concatenation, repetition, indexing, and slicing operations.

The following operations are performed:

1.Create two tuples.
2.Concatenate the two tuples.
3.Repeat the combined tuple three times.
4.Access the third element.
5.Access the first three elements.
6.Access the last three elements.


```python
# Tuples
t1=(10,20,30)
t2=(40,50,60)

# a. Concatenate the two tuples and store it in “t_combine”
t_combine=t1+t2
print(t_combine)

# b. Repeat the elements of “t_combine” 3 times
print(t_combine*3)

# c. Access the 3rd element from “t_combine”
print(t_combine[2])

# d. Access the first three elements from “t_combine”
print(t_combine[:3])

# e. Access the last three elements from “t_combine”
print(t_combine[-3:])
```
